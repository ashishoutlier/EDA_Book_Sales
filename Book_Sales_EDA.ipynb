```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
df = pd.read_csv('Books_Data_Clean.csv')
```

```python
df
```

```python
df.describe()
```

```python
df=df[df['Publishing Year']>1800]
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
df.dropna(subset="Book Name",inplace= True)
```

```python
df.duplicated().sum()
```

```python
df.nunique()
```

```python
plt.hist(df["Publishing Year"]);
plt.xlabel('Publishing Year');
plt.ylabel('Frequency');
plt.title('Distribution of publishing year');
plt.show()
```

```python
df["genre"].value_counts().plot(kind='bar')
plt.xlabel("Genre of book")
plt.ylabel("No of books printed in this genre")
plt.show()
```

```python
sns.boxplot(x= "genre", y = "Book_ratings_count", data = df)
plt.show()
```

```python
plt.scatter(df["sale price"],df["units sold"])
plt.xlabel('Price of book')
plt.ylabel('units of book sold')
plt.show()
```

```python
laguage_counts = df['language_code'].value_counts()
```

```python
plt.pie(laguage_counts, labels=laguage_counts.index,startangle=90)
plt.show()
```
